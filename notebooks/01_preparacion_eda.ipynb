```python
import os
import pandas as pd
import json

# 1. Detección automática de la carpeta (misma lógica que el notebook 02)
if os.path.exists('../data/exercises.json'):
    ruta_data = '../data/'
elif os.path.exists('data/exercises.json'):
    ruta_data = 'data/'
else:
    raise FileNotFoundError("❌ No encuentro exercises.json. Revisa que esté en la carpeta 'data'.")

print(f"✅ Ruta correcta detectada: {ruta_data}")

# 2. Cargar los datos originales
with open(ruta_data + 'exercises.json', 'r', encoding='utf-8') as f:
    datos = json.load(f)
df = pd.DataFrame(datos)

# 3. Funciones de limpieza
def contar_instrucciones_es(x):
    if isinstance(x, dict):
        pasos_es = x.get('es')
        if isinstance(pasos_es, list): return len(pasos_es)
        pasos_en = x.get('en')
        if isinstance(pasos_en, list): return len(pasos_en)
    return 0

def contar_musculos(muscle_group, secondary_muscles):
    total = 0
    if isinstance(muscle_group, str) and muscle_group.strip(): total += 1
    if isinstance(secondary_muscles, list): total += len(secondary_muscles)
    return total

# 4. Feature Engineering
df['num_instrucciones'] = df['instruction_steps'].apply(contar_instrucciones_es) if 'instruction_steps' in df.columns else 0
df['num_musculos_involucrados'] = df.apply(lambda fila: contar_musculos(fila.get('muscle_group'), fila.get('secondary_muscles')), axis=1)
df['instrucciones_texto_es'] = df['instructions'].apply(lambda x: x.get('es', x.get('en', '')) if isinstance(x, dict) else '') if 'instructions' in df.columns else ''
df['musculo_primario'] = df['muscle_group'] if 'muscle_group' in df.columns else ''

# 5. Guardar el dataset limpio
df.to_csv(ruta_data + 'exercises_limpio.csv', index=False)
print(f"¡Éxito! Revisa tu panel izquierdo, exercises_limpio.csv ya debe estar dentro de '{ruta_data}'.")
```
